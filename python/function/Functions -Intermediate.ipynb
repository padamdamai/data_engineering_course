```python
# Example 1: Function to simulate a simple text-based game


def adventure_game():
    """
    A simple text-based game that asks the user to choose between two paths.
    """


    choice = input("You're in a dark forest. Do you go left or right? (left/right): ")
    if choice == "left":
        print("You encounter a dragon!")
        action = input("Do you fight or flee? (fight/flee): ")
        if action == "fight":
            print("You bravely fight the dragon and emerge victorious!")
        else:
            print("You quickly flee and find a safe path home.")
    elif choice == "right":
        print("You find a treasure chest filled with gold!")
    else:
        print("Invalid choice. Game over.")

# Uncomment the line below to play the game (note: this will require user input)
adventure_game()

```

```python

# Example 2: Function to check prime numbers within a range
def print_prime_numbers(start, end):
    """
    This function prints all prime numbers in a given range.
    Parameters:
    start (int): The start of the range.
    end (int): The end of the range.
    """
    for number in range(start, end + 1):

        # Assume number is prime until shown it is not
        is_prime = True

        if number > 1:
            for i in range(2, int(number ** 0.5) + 1):
                if (number % i) == 0:
                    is_prime = False
                    break
            if is_prime:
                print(number, "is a prime number.")
        else:
            # If the number is less than 2, it is not prime
            print(number, "is not a prime number.")

# Example usage
print_prime_numbers(1, 10)
```

```python



# Example 3: Function to convert temperature from Celsius to Fahrenheit and vice versa
def convert_temperature(value, unit):
    """
    This function converts temperature from Celsius to Fahrenheit or vice versa.
    Parameters:
    value (float): The temperature value to convert.
    unit (str): The unit of the input temperature ('C' for Celsius, 'F' for Fahrenheit).

    Returns:
    float: The converted temperature.
    str: The unit of the converted temperature.
    """
    if unit.upper() == 'C':
        converted = (value * 9/5) + 32
        return converted, 'F'
    elif unit.upper() == 'F':
        converted = (value - 32) * 5/9
        return converted, 'C'
    else:
        return "Invalid unit. Use 'C' for Celsius or 'F' for Fahrenheit."

# Example usage
print("25 C is", convert_temperature(25, 'C'), "in Fahrenheit.")
print("77 F is", convert_temperature(77, 'F'), "in Celsius.")
```
